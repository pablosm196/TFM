```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("Resultados.csv", index_col=False)
print(df)
```

# Encuesta SUS

```python
impares = df.iloc[:, [1, 3, 5, 7, 9]].sum(axis = 1) - 5

pares = 25 - df.iloc[:, [2, 4, 6, 8, 10]].sum(axis = 1)

res = (impares + pares) * 2.5
media = res.mean()
dt = res.std()

print(media)
print(dt)
print(res)
```

```python
plt.figure(figsize=(10,6))
plt.bar(df.iloc[:, 0].astype(str).tolist(), res, color = 'steelblue')

plt.title(f"Resultados de la encuesta SUS (N = 3) - μ = {media:.2f}, σ = {dt:.2f}")
plt.xlabel("Usuario")
plt.ylabel("Puntuación")
plt.axhline(y=68, color='red', linestyle='-', linewidth=2, label='Media')

plt.tight_layout()
plt.show()
```

# Tiempos

```python
plt.figure(figsize=(10,6))
plt.bar(df.iloc[:, 0].astype(str).tolist(), df.iloc[:, 11], color = 'steelblue')

plt.title(f"Tiempo de ejecución adecuado (N = 3) - μ = {df.iloc[:, 11].mean():.2f}, σ = {df.iloc[:, 11].std():.2f}")
plt.xlabel("Usuario")
plt.ylabel("Puntuación (1-5)")

plt.tight_layout()
plt.show()
```

```python
users = df.iloc[:,0].astype(str).tolist()

bt1Manual = df.iloc[:, -4].values
bt2Manual = df.iloc[:, -3].values
bt1Herramienta = df.iloc[:, -2].values
bt2Herramienta = df.iloc[:, -1].values

print(f"BT 1 manual: μ = {df.iloc[:, -4].mean()}, σ = {df.iloc[:, -4].std()}")
print(f"BT 1 herramienta: μ = {df.iloc[:, -2].mean()}, σ = {df.iloc[:, -2].std()}")
print(f"BT 2 manual: μ = {df.iloc[:, -3].mean()}, σ = {df.iloc[:, -3].std()}")
print(f"BT 2 herramienta: μ = {df.iloc[:, -1].mean()}, σ = {df.iloc[:, -1].std()}")

n = len(users)

x = np.arange(n)

width = 0.35

plt.figure(figsize=(12, 6))

plt.bar(x - width/2, bt1Manual, width, label="BT1 Manual")
plt.bar(x + width/2, bt1Herramienta, width, label="BT1 Herramienta")

plt.bar(x - width/2 + n + 0.5, bt2Manual, width, label="BT2 Manual")
plt.bar(x + width/2 + n + 0.5, bt2Herramienta, width, label="BT2 Herramienta")

xticks_positions = list(x) + list(x + n + 0.5)
xticks_labels = [f"{u} (1)" for u in users] + [f"{u} (2)" for u in users]

plt.xticks(xticks_positions, xticks_labels, rotation=45)

plt.title("Comparativa de tiempos de generación por pares BT1/BT2 en segundos")
plt.ylabel("Tiempo")
plt.legend()
plt.tight_layout()
plt.show()
```
